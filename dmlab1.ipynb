**Load Datasets**

```python
import pandas as pd
import numpy as np

df = pd.read_csv('employee_data.csv')
print("Initial Data:\n", df.head())
```

****

**Handling missing values**

```python
mean_salary = df['Salary'].mean()
df['Salary'] = df['Salary'].fillna(mean_salary)
mean_age = df['Age'].mean()
df['Age'] = df['Age'].fillna(mean_age)
print(df)
```

**Standardizing department names**

```python
df['Department'] = df['Department'].replace({'HR':'HR','Human Resources':'HR','H.R.':'HR','Finance':'FIN'})
print(df)
```

**remove duplicate record based on id**

```python

df.drop_duplicates(subset=['Name','Age','Department','Salary'], keep='first', inplace=True)
print(df)
```

**data binning1.3**

```python
agedf = pd.read_csv('customer_ages.csv')
print(agedf.head())
```

creating bins fr age column and assigning category abels

```python
bins = [18, 30, 50, 100]
labels = ['Young', 'Middle-aged', 'Senior']

agedf['Age Group'] = pd.cut(agedf['Age'],bins=bins,labels=labels)

print("\n Data Frame after Grouping \n", agedf)
```

calculating number of customer in each age group

```python
agd = agedf['Age Group'].value_counts()
print("\nAge Group Distribution:\n", agd)
```

**Discritization 1.4**

load dataset and inspect first few rows

```python
salesdf = pd.read_csv('sales_data.csv')
print(salesdf.head())
```

applying discritization to sales column

```python
bins = [0, 5000, 20000, float('inf')]
labels = ['Low', 'Medium', 'High']
salesdf['SalesCategory'] = pd.cut(salesdf['Sales'], bins=bins, labels=labels)
print(salesdf)
```

assigning appropriate category label and analyzing the distribution

```python
scd = salesdf['SalesCategory'].value_counts()
print("\nSales Category Distribution:\n", scd)
```

**Feature Selection1.5**

load dataset and inspect first few rows

```python
from sklearn.feature_selection import SelectKBest, chi2
medicaldf = pd.read_csv('medical_data.csv');
print("\n Initial Data sets : \n",medicaldf.head())
```

applying feature selectin method to rank the feature

```python
 X = medicaldf.drop(columns=['Disease'])
 y = medicaldf['Disease']
 selector = SelectKBest(score_func=chi2, k=3)
 selector.fit(X, y)
```

getting top 3 features

```python
tpf = X.columns[selector.get_support()]
print("\nTop 3 Features:\n", tpf)
```
